### 문제 1-3. 수학 수식 (LaTeX) [15점]

Markdown에서 수학 수식은 LaTeX 문법을 사용합니다.

**주요 문법:**
- 합계: `\sum`
- 분수: `\frac{분자}{분모}`
- 위첨자: `x^{2}`
- 아래첨자: `x_{i}`
- 햇(hat): `\hat{x}`
