## 1. 배열 생성 (Creation) #1

Python 리스트와 NumPy 배열의 차이점을 이해합니다.

- Python 리스트: `[1, 2, 3]`
- NumPy 배열: `np.array([1, 2, 3])`

NumPy 배열은 인덱싱 시 `n2[1, 0]` 형태로 접근할 수 있습니다.

---

```python
import numpy as np

## Python 리스트
x1 = [1, 2, 3, 4, 5, 6]
print(x1)

x2 = [[1, 2, 3], [4, 5, 6]]
print(x2)
print(x2[1][0])  # 리스트는 이중 인덱싱 필요

## NumPy 배열로 변환
# IMPLEMENT HERE: x1을 numpy 배열로 변환하여 n1에 저장하세요
n1 = np.array(x1)
print(n1)
print(n1[0])

# IMPLEMENT HERE: x2를 numpy 배열로 변환하여 n2에 저장하세요
n2 = np.array(x2)
print(n2)
print(n2[1, 0])  # NumPy는 콤마로 인덱싱 가능
print(n2[1][0])  # 이중 인덱싱도 가능
```

## 2. 배열 생성 (Creation) #2

NumPy는 특정 값으로 초기화된 배열을 쉽게 생성할 수 있습니다.

- `np.zeros(shape)`: 0으로 채워진 배열
- `np.ones(shape)`: 1로 채워진 배열
- `np.empty(shape)`: 초기화되지 않은 배열 (빠른 생성)
- `dtype` 매개변수로 데이터 타입 지정 가능

---

```python
import numpy as np

# IMPLEMENT HERE: 3x4 크기의 0으로 채워진 배열을 생성하세요
m1 = np.zeros((3,4))
print(m1)
print(m1.dtype)

# IMPLEMENT HERE: 3x4 크기의 1로 채워진 배열을 생성하세요
m2 = np.ones((3,4))
print(m2)
print(m2.dtype)

# dtype을 int16으로 지정한 ones 배열
m3 = np.ones((3, 4), dtype=np.int16)
print(m3)
print(m3.dtype)

# empty 배열 (초기화되지 않음 - 쓰레기 값 포함)
m4 = np.empty((3, 4))
print(m4)
```

## 3. Reshape (배열 형태 변경)

`reshape()` 함수를 사용하여 배열의 형태를 변경할 수 있습니다.

- `np.arange(n)`: 0부터 n-1까지의 배열 생성
- `reshape(행, 열)`: 배열 형태 변경
- `reshape(행, -1)`: -1은 자동 계산

---

```python
import numpy as np

## arange로 배열 생성
r1 = np.arange(6)
print(r1)

# IMPLEMENT HERE: r1을 2행 3열로 reshape하세요
r2 = r1.reshape(2,3)
print(r2)

# -1을 사용한 자동 계산 (2행, 열은 자동)
r3 = np.arange(6).reshape(2, -1)
print(r3)
```

## 4. 기본 연산 (Basic Operation)

NumPy 배열은 요소별(element-wise) 연산을 지원합니다.

- `+`, `-`, `*`, `/`: 요소별 연산
- `**`: 거듭제곱
- `@` 또는 `.dot()`: 행렬 곱셈

**주의:** `*`는 요소별 곱셈, `@`는 행렬 곱셈입니다.

---

```python
import numpy as np

## 기본 연산 1: 요소별 연산
a = np.array([20, 30, 40, 50])
b = np.arange(4)

c1 = a - b
print("a:", a)
print("b:", b)
print("a - b:", c1)

c2 = b ** 2
print("b ** 2:", c2)
```

```python
import numpy as np

## 기본 연산 2: 행렬 곱셈
A = np.array([1, 1, 0, 1]).reshape(2, 2)
B = np.array([2, 0, 3, 4]).reshape(2, 2)

print("A:")
print(A)
print("B:")
print(B)

# 요소별 곱셈
print("A * B (요소별 곱셈):")
print(A * B)

# IMPLEMENT HERE: @ 연산자를 사용하여 행렬 곱셈을 수행하세요
print("A @ B (행렬 곱셈):")
print(A @ B)
print(B @ A)

# IMPLEMENT HERE: dot 메서드를 사용하여 행렬 곱셈을 수행하세요
print("A.dot(B) (행렬 곱셈):")
print(A.dot(B))
```

## 5. Stack (배열 쌓기)

여러 배열을 합칠 수 있습니다.

- `np.hstack()`: 수평으로 쌓기 (가로 방향)
- `np.vstack()`: 수직으로 쌓기 (세로 방향)
- `np.concatenate()`: axis를 지정하여 쌓기

---

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
b = np.array([6, 7, 8, 9, 10])
c = np.array([1, 1, 1, 1, 1])

# IMPLEMENT HERE: a와 b를 수평으로 쌓으세요 (hstack)
o1 = np.hstack((a,b))
print("hstack:", o1)

# IMPLEMENT HERE: a와 b를 수직으로 쌓으세요 (vstack)
o2 = np.vstack((a,b))
print("vstack:")
print(o2)

# a, b, c를 수직으로 쌓기
o3 = np.vstack((a, b, c))
print("vstack (a, b, c):")
print(o3)

# concatenate 사용 (axis=0은 세로 방향)
o4 = np.concatenate((a.reshape(1, -1), b.reshape(1, -1)), axis=0)
# o4 = np.concatenate((a,b), axis=0)
print("concatenate:")
print(o4)

print(a.reshape(1, -1).shape)
```

## 6. Copy (복사)

NumPy 배열의 슬라이싱은 기본적으로 **얕은 복사(shallow copy)**입니다.

- 얕은 복사: 원본과 메모리를 공유 (원본 변경 시 복사본도 변경)
- 깊은 복사: `.copy()` 메서드 사용 (독립적인 복사본 생성)

---

```python
import numpy as np

x = np.arange(10).reshape(2, 5)
print("원본 x:")
print(x)

# 얕은 복사 (shallow copy)
print('\n=== 얕은 복사 ===')
x = np.arange(10).reshape(2, 5)
y = x[0:2, 0:3]
print("y (x의 일부분):")
print(y)

x[0, 0] = 2  # 원본 수정
print("x[0,0] = 2 후 y:")
print(y)  # y도 변경됨!

# 깊은 복사 (deep copy)
print('\n=== 깊은 복사 ===')
x = np.arange(10).reshape(2, 5)
y = x[0:2, 0:3].copy()  # .copy() 사용
print("y (x의 일부분, copy):")
print(y)

x[0, 0] = 2  # 원본 수정
print("x[0,0] = 2 후 y:")
print(y)  # y는 변경되지 않음
```

## 7. Indexing (인덱싱)

NumPy는 불리언 배열을 이용한 인덱싱을 지원합니다.

- `x > 5`: 각 요소가 5보다 큰지 여부를 불리언 배열로 반환
- `x[x > 5]`: 조건을 만족하는 요소만 선택

---

```python
import numpy as np

x = np.array([2, 5, 1, 5, 9, 4, 1, 8, 3, 2])
print("원본 x:", x)

# IMPLEMENT HERE: x에서 5보다 큰 요소를 불리언 배열로 출력하세요
print("x > 5:")
print(x > 5)

# IMPLEMENT HERE: x에서 5보다 큰 요소만 선택하여 출력하세요
print("x[x > 5]:")
print(x[x > 5])

# 조건을 만족하지 않는 요소를 0으로 변경
x[x <= 5] = 0
print("x[x <= 5] = 0 후:", x)
```

## 8. 성능 비교 (Performance Test)

NumPy를 사용하면 Python 리스트보다 훨씬 빠른 연산이 가능합니다.  
아래 코드에서 두 방법의 성능을 비교해봅시다.

---

```python
import numpy as np

# 내적(dot product) 계산의 세 가지 방법
A = list(range(0, 10))
B = list(range(0, 10))

# 방법 1: for 루프 사용
result = 0
for a, b in zip(A, B):
    result += a * b
print("방법 1 (for loop):", result)

# 방법 2: np.dot 사용
result = np.dot(A, B)
print("방법 2 (np.dot):", result)

# 방법 3: @ 연산자 사용
A = np.array(A)
B = np.array(B)
print("방법 3 (@ 연산자):", A @ B)
```

```python
import numpy as np
import time

# 큰 배열에서의 성능 비교
A = list(range(0, 10000))
B = list(range(0, 10000))

# 방법 1: for 루프
result = 0
t0 = time.time()
# IMPLEMENT HERE: for 루프를 사용하여 A와 B의 내적을 계산하세요

t1 = time.time()
print(f'list (for loop): {(t1-t0)*1e3:.4f} ms')

# 방법 2: NumPy
A = np.array(A)
B = np.array(B)
t0 = time.time()
# IMPLEMENT HERE: NumPy의 @ 연산자를 사용하여 내적을 계산하세요

t1 = time.time()
print(f'numpy (@ operator): {(t1-t0)*1e3:.4f} ms')
```
